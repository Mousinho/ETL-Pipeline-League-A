```python
from pyspark.sql.functions import current_timestamp
```

```python
def create_column_date(dataframe):
    dataframe = dataframe.withColumn("ingestion_dt", current_timestamp())
    return dataframe
```

```python
def normalize_columns(df):
    for c in df.columns:
        df = df.withColumnRenamed(c, c.replace(".", "_"))
    return df
```
