### Orchestrator Notebook – Silver Layer Ingestion

**Goal:** Orchestrate the execution of all ingestion notebooks, generating the processed data in the **Silver layer**.

This notebook acts as the entry point, triggering each ingestion notebook according to the dataset — using the **Generic Files Ingestion** for shared files, and **dedicated notebooks** for files with specific transformation logic.

---

### Execution Strategy per File

| File | Notebook | Type |
|---|---|---|
| `championship` | Generic Files Ingestion | Shared generic notebook |
| `stadium` | Generic Files Ingestion | Shared generic notebook |
| `match` | Match Ingestion | Dedicated notebook |
| `team` | Team Ingestion | Dedicated notebook |
| `players` | Players Ingestion | Dedicated notebook |

---

### Execution Flow
```
Orchestrator Notebook
        │
        ├──▶ Generic Files Ingestion (source_file = championship)
        ├──▶ Generic Files Ingestion (source_file = stadium)
        ├──▶ Match Ingestion
        ├──▶ Team Ingestion
        └──▶ Players Ingestion
```

- **Championship** and **Stadium** are processed by the same generic notebook, receiving the source file name via the `source_file` widget
- **Match**, **Team** and **Players** have their own dedicated notebooks due to specific schema or transformation requirements

---

### Important

- The `Generic Files Ingestion` notebook must have a widget named exactly `source_file`
- Each notebook is executed independently via `dbutils.notebook.run`
- A failure in one run does not automatically stop the others unless explicitly handled
