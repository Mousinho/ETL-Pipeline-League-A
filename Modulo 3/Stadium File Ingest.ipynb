```python
path_file_stadium = f"{path_bronze}/estadios.csv"
```

```python
stadium_df = (spark.read
    .option("encoding", "UTF-16")
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(path_file_stadium)
)

display(stadium_df)
stadium_df.printSchema()
```

```python
stadium_df_renamed = stadium_df.withColumnRenamed("id", "stadium_id") \
                               .withColumnRenamed("name", "stadium_name") \
                               .withColumnRenamed("address", "stadium_address") \
                               .withColumnRenamed("city", "stadium_city") \
                               .withColumnRenamed("country", "stadium_country") \
                               .withColumnRenamed("capacity", "stadium_capacity") \
                               .withColumnRenamed("surface", "stadium_surface") \
                               .withColumnRenamed("image", "stadium_image")

display(stadium_df_renamed)
stadium_df_renamed.printSchema()
```

```python
stadium_df_date = create_column_date(stadium_df_renamed)
display(stadium_df_date)
stadium_df_date.printSchema()
```

```python
stadium_df_date.write.mode("overwrite").parquet(f"{path_silver}/estadios")
```
