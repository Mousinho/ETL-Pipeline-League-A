```python
path_file_championship = f"{path_bronze}/torneios.csv"
```

```python
df_championship = spark.read.format("csv") \
.option("header", "true") \
.option("encoding","UTF-16") \
.option("inferSchema", "true") \
.load(path_file_championship)

display(df_championship)
df_championship.printSchema()
```

```python
from pyspark.sql.types import BooleanType

df_championship_type = df_championship.withColumn(
    "season_current", df_championship.season_current.cast(BooleanType())
)
df_championship_type.printSchema()
```

```python
df_championship_date = create_column_date(df_championship_type)
display(df_championship_date)
```

```python
df_championship_date.write.mode("overwrite").parquet(f"{path_silver}/torneios")
```
