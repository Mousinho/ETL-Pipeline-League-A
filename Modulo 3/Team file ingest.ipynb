```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    BooleanType,
    DoubleType,
)

schema_team = StructType(
    fields=[
        StructField("id", StringType(), False),
        StructField("name", StringType(), False),
        StructField("code", StringType(), False),
        StructField("country", StringType(), False),
        StructField("founded", DoubleType(), False),
        StructField("national", BooleanType(), False),
        StructField("logo", StringType(), False)
    ]
)
```

```python
path_team = f'{path_bronze}/times.json'
```

```python
df_team = spark.read.json(path_team)
display(df_team)
```

```python
df_team = spark.read.schema(schema_team).json(path_team)

display(df_team)
df_team.printSchema()
```

```python
df_team = df_team.withColumn('founded', df_team.founded.cast('integer'))
df_team.printSchema()
```
