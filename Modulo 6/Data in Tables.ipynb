```python
players_df.createOrReplaceTempView("view_players")
```

```python
%sql
SELECT * FROM view_players
```

```python
%sql
select
  distinct id as id_player,
  name,
  DATEDIFF(YEAR, birth_date, GETDATE()) as age,
  concat(coalesce(birth_place,'Not Defined' ), ', ', birth_country) as birth_place,
  birth_country as country,
  weight,
  height,
  photo
  from
  view_players
```

```python
%sql
CREATE OR REPLACE TABLE futdata.dim_players
AS
select
  distinct id as id_player,
  name,
  DATEDIFF(YEAR, birth_date, GETDATE()) as age,
  concat(coalesce(birth_place,'Not Defined' ), ', ', birth_country) as birth_place,
  birth_country as country,
  weight,
  height,
  photo
from
  view_players
```
