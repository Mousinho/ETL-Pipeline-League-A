```sql
SELECT
  referee,
  COUNT(1) AS n_matches
FROM
  futdata.ft_matches
WHERE
  league_season = 2022
GROUP BY
  referee
ORDER BY
  n_matches DESC
```

```sql
SELECT
 Name_Location,
 SUM(goals_home) + sum(goals_away) as total_goals
FROM futdata.ft_matches
GROUP BY ALL
ORDER BY 2 DESC
```
