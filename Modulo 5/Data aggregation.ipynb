```python
matches_df = spark.read.parquet(f"{path_gold}/fact_matches")
display(matches_df)
```

```python
from pyspark.sql.functions import count, countDistinct, max, min, sum, avg
```

```python
matches_df.select(count('*')).display()

matches_df.select(countDistinct('referee')).display()
```

```python
matches_df.select('referee').distinct().display()
```

```python
df_goals = (
    matches_df.filter("league_round = 'Regular Season - 1'")
    .select(sum("goals_home"), sum("goals_away"))
    .withColumnRenamed("sum(goals_home)", "total_goals_home")
    .withColumnRenamed("sum(goals_away)", "total_goals_away")

)

display(df_goals)
```

```python
matches_df.groupBy('league_season').sum('goals_home').display()
```

```python
matches_df.groupBy("league_season").agg(
    sum("goals_home").alias('Total_goals_home'), 
    sum("goals_away").alias('Total_goals_away'), 
    avg("goals_home").alias('avg_goals_home'), 
    avg("goals_away").alias('avg_goals_away')
).display()
```
