```python
matches_df = spark.read.parquet(f"{path_gold}/fact_matches")
```

```python
df_reduced = matches_df.select("league_season", "teams_home_id", "goals_home")
display(df_reduced)
```

```python
df_group = df_reduced.groupBy('league_season', 'teams_home_id').sum('goals_home')
display(df_group)
```

```python
from pyspark.sql.window import Window
```

```python
from pyspark.sql.functions import rank, dense_rank, row_number, desc
```

```python
window = Window.partitionBy('league_season').orderBy(desc('sum(goals_home)'))
```

```python
df_rank = df_group.withColumn('rank', dense_rank().over(window))
display(df_rank)
```

```python
df_row_num_number = df_group.withColumn('r_n', row_number().over(window))
display(df_row_num_number)
```

```python
df_row_num_number.filter("r_n = 1").display()
```
