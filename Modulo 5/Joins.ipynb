### Join of Matches and Stadiums Tables
1. Read the tables
2. Filter only the rows related to matches
3. Perform a join between the tables
4. Create a new column with the city and country name
5. Save the final table in the gold layer

```python
file_name = "estadios"
path = f"{path_silver}/{file_name}"
stadium_df = spark.read.parquet(path)
```

```python
path_file_match = f"{path_silver}/partidas/"

match_df = (
    spark.read
    .parquet(path_file_match)
)

display(match_df)
```

```python
print(f"Exist {match_df.count()} matches in the dataset.")
print(f"Exist {stadium_df.count()} stages in the dataset.")
```

```python
display(match_df)
display(stadium_df)
```

```python
matches_stadium_df = match_df.join(
    stadium_df,
    on=match_df.venue_id == stadium_df.stadium_id,
    how="left"
)
display(matches_stadium_df)
```

```python
matches_stadium_df = matches_stadium_df.select (
    match_df['*'],
    stadium_df['stadium_city'],
    stadium_df['stadium_country']
)
```

```python
display(matches_stadium_df)
```

```python
from pyspark.sql.functions import concat, col

matches_stadium_df = matches_stadium_df.withColumn(
    "Name_Location", concat(col("stadium_city"), col("stadium_country"))
    )
```

```python
display(matches_stadium_df)
```

```python
table_name = "fact_matches"
output_path = f"{path_gold}/{table_name}"

matches_stadium_df.write.mode("overwrite").parquet(output_path)

print(f"Data written to {output_path}")
```
