1. Creating a dataframe manually


```python
from pyspark.sql.types import StructType, StructField, StringType, DateType
```

```python
Schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Function", StringType(), True),
    StructField("Date", StringType(), True)
])

data = [
    ("Pedro", "Data Engineer", "01/04/2026"),
    ("Paula", "Social Worker", "05/09/2026"),
    ("Vitor", "Student", "01/02/2020")
    ]

df = spark.createDataFrame(data, schema= Schema)
```

```python
display(df)
```

2. Converting text to date

```python
from pyspark.sql.functions import to_date
```

```python
df = df.withColumn("Date", to_date(df["Date"], "dd/MM/yyyy"))
```

```python
display(df)
df.printSchema()
```

2.1 Timestamp function

```python
from pyspark.sql.functions import to_timestamp
```

```python
df = df.withColumn("Date", to_timestamp(df["Date"], "dd/MM/yyyy"))
```

```python
display(df)
df.printSchema()
```

3. Transforming into text

```python
from pyspark.sql.functions import date_format
```

```python
df = df.withColumn("Date", date_format(df["Date"], "dd-MM-yyyy"))
```

```python
display(df)
df.printSchema()
```

4. Calculations with date

```python
Schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Function", StringType(), True),
    StructField("Date_B", StringType(), True),
    StructField("Date_F", StringType(), True)
])

data = [
    ("Pedro", "Data Engineer", "01/04/2026", "01/04/2050"),
    ("Paula", "Social Worker", "05/09/2026", "01/04/2046"),
    ("Vitor", "Student", "01/02/2025", "01/04/2025")
    ]

df2 = spark.createDataFrame(data, schema= Schema)
```

```python
df2 = df2 \
    .withColumn("Date_B", to_date(df2["Date_B"], "dd/MM/yyyy")) \
    .withColumn("Date_F", to_date(df2["Date_F"], "dd/MM/yyyy"))
```

```python
display(df2)
```

```python
from pyspark.sql.functions import col, datediff, date_add, date_sub
```

```python
df_dff = df2 \
    .withColumn("Diff_date", datediff(col('Date_F'), col('Date_B'))) \
    .withColumn("Add_date", date_add(col('Date_B'), 7)) \
    .withColumn("Sub_date", date_sub(col('Date_B'), 3))
```

```python
display(df_dff)
```

5. Extracting components from dates

```python
from pyspark.sql.functions import year, month, dayofmonth, hour, minute, second
```

```python
Schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Function", StringType(), True),
    StructField("Date_B", StringType(), True)
])

data = [
    ("Pedro", "Data Engineer", "01/04/2026 10:00:00"),
    ("Paula", "Social Worker", "05/09/2026 15:41:10"),
    ("Vitor", "Student", "01/02/2025 23:30:30")
    ]

df3 = spark.createDataFrame(data, schema= Schema)
```

```python
df3 = df3.withColumn("Date_B", to_timestamp(df3["Date_b"], "dd/MM/yyyy HH:mm:ss"))
display(df3)
```

```python
df3 = df3 \
    .withColumn("year", year(col('Date_B'))) \
    .withColumn("month", month(col('Date_B'))) \
    .withColumn("dayofmonth", dayofmonth(col('Date_B'))) \
    .withColumn("hour", hour(col('Date_B'))) \
    .withColumn("minute", minute(col('Date_B'))) \
    .withColumn("second", second(col('Date_B')))
```

```python
display(df3)
```

6. Filtering Dates

```python
display(df3.filter(df3["Date_B"] >= "2026-01-01"))
```
